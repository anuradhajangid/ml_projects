```python
import faiss
import numpy as np

# Create a random vector database of 1000 vectors, each of dimension 128
d = 128  # Dimension of vectors
nb = 1000  # Number of vectors
nq = 10  # number of queries
xb = np.random.random((nb, d)).astype('float32')
xq = np.random.random((nq, d)).astype('float32')

# Create a FAISS index
index = faiss.IndexFlatL2(d)  # L2 distance (Euclidean)
index.add(xb)  # Add vectors to the index

print("Index created successfully!")
```

```python
k = 5  # number of nearest neighbors
D, I = index.search(xq, k)

print("Distances:", D)
print("Indices:", I)
```

```python
import faiss
import numpy as np
from sentence_transformers import SentenceTransformer

# Initialize a pre-trained sentence transformer model
model = SentenceTransformer('all-MiniLM-L6-v2')

# Example sentences
sentences = ["This is a sentence.", "FAISS is a library for vector search.", "Machine learning is fascinating."]

# Convert sentences to embeddings (vectors)
embeddings = model.encode(sentences)

# Convert embeddings to numpy array of type float32 (required by FAISS)
embeddings = np.array(embeddings).astype('float32')

# Create a FAISS index
d = embeddings.shape[1]  # Embedding dimension
index = faiss.IndexFlatL2(d)  # L2 distance (Euclidean)

# Add the embeddings to the index
index.add(embeddings)

# Query example (for example, the first sentence)
query = model.encode(["FAISS library example"]).astype('float32')

# Search for nearest neighbors (top 3 neighbors)
k = 3
distances, indices = index.search(query, k)

print("Nearest Neighbors (Indices):", indices)
print("Distances:", distances)
```
